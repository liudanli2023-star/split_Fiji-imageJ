# 批量重命名 tile 图像

遍历指定组别和通道文件夹，将其中的 .tif 文件按排序顺序重命名为 tile_01.tif、tile_02.tif 等标准名称。

注释说明：
- os.rename 会直接改名，运行前建议确认文件排序符合真实 tile 顺序。
- 如果目标文件名已存在，可能触发覆盖/冲突风险，建议在备份目录中测试。


```python
import os

# 1. 定义大组（A, B, C, D）
#groups = ["A", "B", "C", "D"]
groups = ["A"]

# 2. 定义每一组下属的 5 个通道文件夹
channels = ["488", "546", "647", "cy7", "DAPI"]

# 3. 基础路径 (在 Python 中，路径要么用双反斜杠 '\\'，要么在字符串前加 'r')
base_path = r"D:\01.analysis\11.test_result\P4-rep2-A-merge"

print("🚀 开始遍历目录并进行重命名...")
print("-" * 50)
```

```python
# 4. 嵌套循环遍历多级目录
for g in groups:
    for ch in channels:
        # 动态拼接出目标文件夹的完整路径
        # 例如：F:\dir\P4\A\P4-rep2-A-stack\488
        target_path = os.path.join(base_path, g, f"P4-rep2-{g}-stack", ch)
        
        # 检查该路径在电脑中是否存在
        if os.path.exists(target_path):
            print(f"📁 正在处理目录: {target_path}")
            
            # 获取该文件夹下所有的文件
            all_files = os.listdir(target_path)
            
            # 过滤出所有 .tif 文件，并按原文件名排序（确保 tile_01 对应 Ch1 递增顺序）
            tif_files = [f for f in all_files if f.lower().endswith('.tif')]
            tif_files.sort()
            
            count = 0
            for file_name in tif_files:
                # 计数器从 1 开始
                count += 1
                
                # 自动补零：{:02d} 会把 1 变成 "01"，2 变成 "02"
                # 如果总文件数超过 100 个，可以改为 {:03d}
                new_name = f"tile_{count:02d}.tif"
                
                # 构建完整的旧路径与新路径
                old_file_path = os.path.join(target_path, file_name)
                new_file_path = os.path.join(target_path, new_name)
                
                # 执行重命名
                os.rename(old_file_path, new_file_path)
                
            print(f"   └─ 重命名完成！该文件夹共处理了 {count} 个文件。")
        else:
            # 如果某个路径不存在（例如没有D组），则跳过打印提示
            print(f"⚠️ 路径不存在，已跳过 -> {target_path}")

print("-" * 50)
print("🎉 所有目录全部处理完毕！")
```
