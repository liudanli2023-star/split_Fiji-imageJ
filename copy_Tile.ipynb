# 复制 TileConfiguration 到各荧光通道

在每个 stage/replicate/letter 样本中，把 DAPI 文件夹下的 TileConfiguration.txt 复制到 488、546、647、cy7 等通道文件夹，便于对各通道使用相同拼接坐标。

注释说明：
- 运行前确认 DAPI 坐标文件已经生成。
- target_subfolders 可按实际通道名调整。


```python
import shutil
from pathlib import Path

# 1. 定义根路径
base_path = Path(r"D:\LDL_data\01.analysis\02.merge")

# 2. 定义变量范围
stages = ["e18.5"]
replicates = ["rep1", "rep2"]
letters = [chr(i) for i in range(ord('A'), ord('J') + 1)]  # A 到 J

# 3. 目标通道文件夹
target_subfolders = ["488", "546", "647", "cy7"]

print("🚀 开始精确匹配并复制 TileConfiguration.txt ...")

# 嵌套循环处理所有组合
for stage in stages:
    for rep in replicates:
        for letter in letters:
            
            # 当前字母根目录，例如：D:\...\e18.5\B
            current_folder = base_path / stage / letter
            
            # 样本文件夹名，例如：e18.5-rep2-B
            sample_folder_name = f"{stage}-{rep}-{letter}"
            
            # 修正后的源文件路径：.../e18.5-rep2-B/DAPI/TileConfiguration.txt
            source_file = current_folder / sample_folder_name / "DAPI" / "TileConfiguration.txt"
            
            # 如果该样本或对应的 TileConfiguration.txt 不存在，则跳过
            if not source_file.exists():
                continue

            print(f"📂 正在处理: {sample_folder_name}")

            # 循环复制到对应的通道文件夹中
            for subfolder in target_subfolders:
                # 目标目录，例如：D:\...\e18.5\B\e18.5-rep2-B\488
                dest_dir = current_folder / sample_folder_name / subfolder
                
                # 确保目标通道文件夹存在（不存在则自动创建）
                dest_dir.mkdir(parents=True, exist_ok=True)
                
                # 目标文件完整路径
                dest_file = dest_dir / "TileConfiguration.txt"
                
                # 执行复制
                shutil.copy2(source_file, dest_file)
                print(f"  └─ 成功复制到 -> {subfolder}/")

print("\n✨ 所有指定通道的 TileConfiguration.txt 均已成功复制并对齐！")
```
