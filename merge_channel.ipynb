# 简易多通道 TIFF 合并

读取五个 fused 单通道 TIFF，堆叠为 C-Y-X 结构并保存为一个多通道 TIFF。

注释说明：
- 相比 03.channel.ipynb，该脚本更简洁，不写入完整 OME 通道颜色元数据。


```python
import tifffile as tiff
import numpy as np
import os
```

```python
input_dir = r"D:\01.analysis\11.test_result\P4-rep2-A-merge\A\P4-rep2-A-stack"
```

```python
imgs = [
    tiff.imread(os.path.join(input_dir, "DAPI-Fused.tif")),
    tiff.imread(os.path.join(input_dir, "488-Fused.tif")),
    tiff.imread(os.path.join(input_dir, "546-Fused.tif")),
    tiff.imread(os.path.join(input_dir, "647-Fused.tif")),
    tiff.imread(os.path.join(input_dir, "cy7-Fused.tif")) 
    
]

merged = np.stack(imgs)
```

```python
output_file = r"D:\01.analysis\11.test_result\P4-rep2-A-merge\A\P4-rep2-A-stack\P4-rep2-A.tif"
```

```python
tiff.imwrite(
    output_file,
    merged,
    metadata={"axes": "CYX"}
)

print(f"保存完成: {output_file}")
print(f"图像维度: {merged.shape}")
```
